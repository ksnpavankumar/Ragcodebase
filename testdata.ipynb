```python
import os
import pandas as pd
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer
from reportlab.lib.styles import getSampleStyleSheet

# 1. Generate Sample Jira Tickets (CSV)
jira_data = {
    "Issue Key": ["ENG-101", "ENG-102", "FIN-204"],
    "Status": ["Done", "In Progress", "Open"],
    "Assignee": ["Alice Smith", "Bob Jones", "Charlie Dev"],
    "Summary": [
        "Implement OAuth2 token rotation for banking microservices",
        "Fix database connection pool timeout during high concurrency",
        "Add audit logging for international wire transfers"
    ],
    "Description": [
        "Ensure refresh tokens expire after 7 days and are securely stored in Redis. Referenced in ENG-102 architecture design.",
        "HikariCP pool gets exhausted during peak banking hours. Increase maximum pool size and reduce connection timeout limits.",
        "All transactions exceeding $10,000 must trigger an asynchronous audit event logged to the compliance datastore. Linked to compliance guidelines in banking PDF."
    ]
}
df_jira = pd.DataFrame(jira_data)
df_jira.to_csv("sample_jira_tickets.csv", index=False)

# 2. Generate Sample Confluence Export (PDF)
confluence_pdf_path = "sample_confluence_export.pdf"
doc1 = SimpleDocTemplate(confluence_pdf_path, pagesize=letter)
styles1 = getSampleStyleSheet()
story1 = [
    Paragraph("<b>Confluence Documentation: Authentication & Security Architecture</b>", styles1['Heading1']),
    Spacer(1, 12),
    Paragraph("This document outlines core security practices for internal microservices. All authentication flows pass through our centralized gateway.", styles1['Normal']),
    Spacer(1, 10),
    Paragraph("<b>Related Tickets & Tracking:</b> Token implementation details are tracked under ENG-101. Engineers must ensure backward compatibility.", styles1['Normal']),
    Spacer(1, 10),
    Paragraph("<b>Database Pooling:</b> To prevent downstream failures, connection thresholds are strictly managed as outlined in system configurations.", styles1['Normal'])
]
doc1.build(story1)

# 3. Generate Sample Banking Knowledge (PDF)
banking_pdf_path = "sample_banking_knowledge.pdf"
doc2 = SimpleDocTemplate(banking_pdf_path, pagesize=letter)
styles2 = getSampleStyleSheet()
story2 = [
    Paragraph("<b>Enterprise Banking Compliance & Wire Transfer Policy v4.2</b>", styles2['Heading1']),
    Spacer(1, 12),
    Paragraph("Section 1.1: Regulatory Compliance. All domestic and international fund transfers are subject to strict validation frameworks.", styles2['Normal']),
    Spacer(1, 10),
    Paragraph("Section 1.2: Audit Controls. High-value transactions require mandatory logging and secondary verification checks. Please refer to ticket FIN-204 for ongoing engineering implementation tracking.", styles2['Normal']),
    Spacer(1, 10),
    Paragraph("Section 1.3: Data Retention. Compliance logs must be retained for a minimum duration of 7 years in an immutable cold storage tier.", styles2['Normal'])
]
doc2.build(story2)

print("✅ Mock files generated successfully: sample_jira_tickets.csv, sample_confluence_export.pdf, sample_banking_knowledge.pdf")
```
